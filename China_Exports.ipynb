```python
import pandas as pd

database = pd.read_csv("China_Exports.csv", sep=';')

#Stripping leading and trailing whitespace and control characters from column names and text fields
database.columns = database.columns.str.strip()
database[["Country", "City","SH4 Description"]] = database[["Country", "City","SH4 Description"]].apply(lambda x: x.str.strip())
#Extracting the month name by splitting the string on the period, selecting the second element, and stripping whitespace
database["Month"] = database["Month"].str.split(".").str[1].str.strip()

display(database.head(10))
```

```python
database.info()
```

```python
#Analysing each month value and formating it
export_amount_months = database[["Month","US$ FOB"]].groupby("Month").sum(numeric_only=True)

def format(value):
    format_value = f"US$ {value:,}"
    return format_value

export_amount_months["US$ FOB"] = export_amount_months["US$ FOB"].apply(format)


display(export_amount_months)
```

```python
#Most exported products troughout the months

export_amount_months = database[["SH4 Description","US$ FOB"]].groupby("SH4 Description").sum()
export_amount_months = export_amount_months.sort_values(by="US$ FOB", ascending=False)

export_amount_months["US$ FOB"] = export_amount_months["US$ FOB"].apply(format)
display(export_amount_months.head(6))
```

```python
#The city with more exports to China in March

table_march = database.loc[database["Month"]=="March",:]
table_march = table_march[["City","SH4 Description","US$ FOB"]].groupby("City").sum()
table_march = table_march.sort_values(by="US$ FOB",ascending=False)
table_march["US$ FOB"] = table_march["US$ FOB"].apply(format)
table_march = table_march.reset_index()

display(table_march)


```

```python
print(table_march.columns)
```

```python
#Automating tha analysis by city



city = table_march.loc[0, "City"]   

table_city = database.loc[(database["Month"]=="March") & (database["City"]==city), :]

table_products_city = table_city[["SH4 Description","US$ FOB"]].groupby("SH4 Description").sum()
table_products_city = table_products_city.sort_values(by="US$ FOB",ascending=False)

table_products_city["US$ FOB"] = table_products_city["US$ FOB"].apply(format)

print(f"The most exported product by {city}:")
display(table_products_city)
```
